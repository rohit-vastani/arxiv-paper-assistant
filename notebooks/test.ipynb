```python
from langchain.mcp import MCPAdapter
```

```python
CONFIG = {
    "mcpServers":{
        "search_and_download": {"command": "python", "args": ["/home/rohit_vastani/academic-mcp-agent/mcp_server/arxiv_search.py"]},
        "cread_store_vec": {"command": "python", "args": ["/home/rohit_vastani/academic-mcp-agent/mcp_server/store_vec.py"]},
        "context_retriever": {"command": "python", "args": ["/home/rohit_vastani/academic-mcp-agent/mcp_server/search_context.py"]}
    }
}
```

```python
async with MCPAdapter(CONFIG) as adapter:
    tools = await adapter.list_tools()
```

```python
tools
```

```python
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver
from langchain.agents import create_agent
from dotenv import load_dotenv; load_dotenv()
```

```python
system_prompt = """You are a helpful AI assistant which explain the user about research paper.
You have access to several tools so that you can download research paper.
After donwloading the research paper you can store it into the vector database and get the relevant content from it related to the user query.
When the user give the topic name or intended to topic then only donwload the research paper and then process it.
After that as user ask the question get the context from the tools and explain to the user.
"""
```

```python
agent = create_agent(
    model="google_genai:gemini-3.1-flash-lite",
    tools=tools,
    system_prompt=system_prompt,
    checkpointer=InMemorySaver(),
)
```

```python
config = {"configurable": {"thread_id": "1"}}
```

```python
response = await agent.ainvoke(
    {"messages": [HumanMessage(content="Can you help me to understand some research papers?")]},
    config=config
)
```

```python
print(response["messages"][-1].content[0]["text"])
```
